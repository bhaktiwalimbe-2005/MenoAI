```python
from datetime import datetime, timedelta
import csv
import os

# --------- Menstrual Cycle Tracker with CSV Saving ---------
def track_menstrual_cycle(last_period_date_str, cycle_length=28):
    try:
        last_period_date = datetime.strptime(last_period_date_str, "%Y-%m-%d")
    except ValueError:
        print("Invalid date format. Please use YYYY-MM-DD.")
        return

    next_period_date = last_period_date + timedelta(days=cycle_length)
    ovulation_day = last_period_date + timedelta(days=cycle_length - 14)
    fertile_start = ovulation_day - timedelta(days=4)
    fertile_end = ovulation_day + timedelta(days=1)

    # Display details
    print("\n📅 Menstrual Cycle Details:")
    print(f"Last Period Start Date : {last_period_date.date()}")
    print(f"Next Period Expected   : {next_period_date.date()}")
    print(f"Ovulation Day          : {ovulation_day.date()}")
    print(f"Fertile Window         : {fertile_start.date()} to {fertile_end.date()}")

    # Save to CSV
    file_exists = os.path.isfile("cycle_tracker.csv")
    with open("cycle_tracker.csv", "a", newline="") as file:
        writer = csv.writer(file)
        if not file_exists:
            writer.writerow(["Last_Period_Start", "Next_Period", "Ovulation_Day",
                             "Fertile_Window_Start", "Fertile_Window_End", "Cycle_Length"])
        writer.writerow([last_period_date.date(), next_period_date.date(), ovulation_day.date(),
                         fertile_start.date(), fertile_end.date(), cycle_length])

# --------- Symptom Tracker ---------
def log_symptom(symptom, severity):
    today = datetime.now().date()
    file_exists = os.path.isfile("symptom_tracker.csv")
    with open("symptom_tracker.csv", "a", newline="") as file:
        writer = csv.writer(file)
        if not file_exists:
            writer.writerow(["Date", "Symptom", "Severity"])
        writer.writerow([today, symptom, severity])
    print(f"✅ Symptom '{symptom}' logged with severity '{severity}' on {today}")

# --------- AI Healthcare Chatbot ---------
from transformers import pipeline

qa_model = pipeline("question-answering", model="deepset/roberta-base-squad2")

medical_knowledge = {
    "What are common symptoms of periods?": "Cramps, bloating, breast tenderness, mood swings, fatigue, headaches, and acne.",
    "How can I reduce period cramps?": "Warm compresses, gentle exercise, staying hydrated, or using over-the-counter pain relief like ibuprofen can help.",
    "Is it normal to have irregular periods?": "Yes, especially in the first couple of years after menstruation begins or due to stress, diet, or hormonal changes.",
    "How often should I change my pad or tampon?": "Every 4–6 hours or sooner if it's full to prevent odor and reduce the risk of infections.",
    "Can I bathe or shower during my period?": "Yes, at least one time a day.",
    "What is Toxic Shock Syndrome (TSS)?": "A rare but serious infection linked to prolonged tampon use. Always change tampons regularly.",
    "Why do I feel emotional before my period?": "Hormonal changes before your period can lead to PMS, which includes mood swings, irritability, or sadness.",
    "How can I manage mood swings during periods?": "Exercise, sleep, reducing sugar intake, and practicing mindfulness or relaxation techniques can help.",
    "What should I eat during my period?": "Foods rich in iron, magnesium, and fiber such as leafy greens, bananas, whole grains, and lean meats help. Avoid too much salt, sugar, and caffeine.",
    "Is exercise safe during periods?": "Yes! Light activities like yoga, walking, or stretching can relieve cramps and improve mood.",
    "Can I go swimming during my period?": "Yes, using a tampon or menstrual cup makes swimming hygienic and safe.",
    "Is period blood dirty?": "No, it’s just the shedding of the uterine lining — it’s completely natural and not dirty.",
    "Can I get pregnant during my period?": "It’s unlikely but possible, especially with irregular cycles. Sperm can live for up to 5 days in the body.",
    "When should I be concerned about my period?": "See a doctor if you experience very heavy bleeding, periods lasting over 7 days, no period for over 3 months, or severe pain not relieved by medication.",
    "Are there medical conditions related to menstruation?": "Yes, conditions like PCOS, endometriosis, or thyroid disorders can affect menstrual health."
}

def get_answer(question):
    if question in medical_knowledge:
        return medical_knowledge[question]
    else:
        best_match = ""
        best_score = 0
        for key in medical_knowledge:
            result = qa_model(question=question, context=medical_knowledge[key])
            if result["score"] > best_score:
                best_score = result["score"]
                best_match = result["answer"]
        return best_match if best_score > 0.5 else "I'm sorry, I don't have enough information to answer your question. Please consult a healthcare professional."

# --------- Main Program ---------
if __name__ == "__main__":
    print("🌸 Menstrual Date Tracker 🌸")
    user_input = input("Enter your last period start date (YYYY-MM-DD): ")
    cycle_input = input("Enter your average cycle length in days (default 28): ")
    cycle_length = 28 if cycle_input.strip() == "" else int(cycle_input)

    # Track cycle and save
    track_menstrual_cycle(user_input, cycle_length)

    # Symptom logging option
    while True:
        log_choice = input("Do you want to log a symptom? (yes/no): ").lower()
        if log_choice == "yes":
            symptom = input("Enter symptom: ")
            severity = input("Enter severity (Mild/Moderate/Severe): ")
            log_symptom(symptom, severity)
        else:
            break

    # AI chatbot
    print("\n🤖 AI Healthcare Chatbot (Type 'exit' to quit)")
    while True:
        user_question = input("You: ")
        if user_question.lower() in ["exit", "quit", "bye"]:
            print("Chatbot: Take care! Stay healthy! 😊")
            break
        answer = get_answer(user_question)
        print("Chatbot:", answer)
```
